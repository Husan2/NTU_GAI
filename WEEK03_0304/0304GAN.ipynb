## 我對於GAN的觀點：
- GAN的生成的圖片在手指表現上還有待加強，另外在配色的部分也沒有完全按照指令運行。

- 生成多樣性較低：GAN 容易模式崩潰，無法像 Diffusion Models 一樣生成更具創意的圖片。

# 3.使用 DALL·E 3 生成圖片

# 結論：
| 模型          | 優點 | 缺點 |
|--------------|-----|-----|
| **StyleGAN2-ADA** | 高解析度、高品質人臉生成 | 訓練成本高、不適用於文本輸入 |
| **DALL·E** | 文字轉圖像能力強、適用於多種風格 | 計算成本高、細節有時失真 |
| **Stable Diffusion** | 低成本、可本地運行、支持文字提示 | 需要額外調整參數以生成高品質圖像 |

雖然 GAN 仍然適用於某些特定領域 (如人臉生成、物件重建)，但在 多樣性、控制能力、運行成本 上，它已經被 DALL·E 和 Stable Diffusion 超越。因此，現在 DALL·E 和 Stable Diffusion 是 AI 圖像生成的主流選擇。
