# 建立 Streamlit 應用程式

```python
import streamlit as st
import time
import threading
import inspect  # 新增這行導入 inspect 模組

# 貓咪風格的番茄鐘計時器
class CatTomatoTimer:
    def __init__(self):
        # 初始化會話狀態變量
        if 'timer_running' not in st.session_state:
            st.session_state.timer_running = False
        if 'remaining_time' not in st.session_state:
            st.session_state.remaining_time = 25 * 60
        if 'is_working' not in st.session_state:
            st.session_state.is_working = True
        if 'completed_pomodoros' not in st.session_state:
            st.session_state.completed_pomodoros = 0

    def start_timer(self):
        st.session_state.timer_running = True

    def pause_timer(self):
        st.session_state.timer_running = False

    def reset_timer(self):
        st.session_state.timer_running = False
        st.session_state.remaining_time = 25 * 60
        st.session_state.is_working = True

    def countdown(self):
        if st.session_state.timer_running and st.session_state.remaining_time > 0:
            time.sleep(1)
            st.session_state.remaining_time -= 1

def main():
    # 設置頁面標題和圖標
    st.set_page_config(
        page_title="貓咪番茄鐘",
        page_icon="🍅",
        layout="centered"
    )

    # 自定義 CSS 樣式
    st.markdown("""
    <style>
    .stApp {
        background-color: #FFF0F5;  /* 淡粉色背景 */
    }
    .title {
        color: #FF6347;
        text-align: center;
        font-family: 'Comic Sans MS', cursive;
    }
    .timer {
        font-size: 3em;
        color: #FF4500;
        text-align: center;
        font-family: 'Comic Sans MS', cursive;
    }
    .status {
        text-align: center;
        font-size: 1.5em;
        color: #FF6347;
    }
    </style>
    """, unsafe_allow_html=True)

    # 標題
    st.markdown('<h1 class="title">🍅 貓咪番茄鐘 🐱</h1>', unsafe_allow_html=True)

    timer = CatTomatoTimer()

    # 顯示時間
    mins, secs = divmod(st.session_state.remaining_time, 60)
    time_display = f"{mins:02d}:{secs:02d}"
    st.markdown(f'<div class="timer">{time_display}</div>', unsafe_allow_html=True)

    # 狀態顯示
    if st.session_state.is_working:
        st.markdown('<div class="status">🧑‍💻 專注工作中</div>', unsafe_allow_html=True)
    else:
        st.markdown('<div class="status">😴 休息時間</div>', unsafe_allow_html=True)

    # 顯示已完成的番茄週期
    st.write(f"已完成番茄週期: {st.session_state.completed_pomodoros}")

    # 按鈕列
    col1, col2, col3 = st.columns(3)

    with col1:
        if st.button("🟢 開始", key="start"):
            timer.start_timer()

    with col2:
        if st.button("🔴 暫停", key="pause"):
            timer.pause_timer()

    with col3:
        if st.button("🔄 重置", key="reset"):
            timer.reset_timer()

    # 客製化工作和休息時間
    st.sidebar.header("⚙️ 設定")
    work_minutes = st.sidebar.slider("工作時間（分鐘）", 20, 45, 25)
    break_minutes = st.sidebar.slider("休息時間（分鐘）", 3, 15, 5)

    # 後台倒計時
    if st.session_state.timer_running:
        timer.countdown()
        st.experimental_rerun()

    # 時間到提醒
    if st.session_state.remaining_time <= 0:
        if st.session_state.is_working:
            st.balloons()
            st.success("🎉 工作時間結束！開始休息")
            st.session_state.is_working = False
            st.session_state.remaining_time = break_minutes * 60
            st.session_state.completed_pomodoros += 1
        else:
            st.balloons()
            st.success("🎉 休息結束！準備下一個工作週期")
            st.session_state.is_working = True
            st.session_state.remaining_time = work_minutes * 60
```

```python
# 將應用程式儲存為 Python 檔案
with open('/content/tomato_timer_app.py', 'w') as f:
    f.write('import streamlit as st\n')  # 新增必要的導入
    f.write('import time\n')
    f.write('import threading\n\n')
    f.write(inspect.getsource(CatTomatoTimer))
    f.write('\n\n')
    f.write(inspect.getsource(main))
    f.write('\n\nif __name__ == "__main__":\n    main()')
```
