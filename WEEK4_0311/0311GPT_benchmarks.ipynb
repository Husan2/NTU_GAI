# 🐱 三、建立 Flask 應用程式

```python
%%writefile app.py
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def home():
    return render_template("index.html")

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)
```

# 🐱 四、建立 HTML 模板與UI介面

```python
!mkdir templates
```

在 templates 中建立index.html：

```python
%%writefile templates/index.html
<!DOCTYPE html>
<html lang="zh-TW">
<head>
    <meta charset="UTF-8">
    <title>貓咪蕃茄鐘</title>
    <link rel="stylesheet" href="https://cdnjs.cloudflare.com/ajax/libs/twitter-bootstrap/5.3.3/css/bootstrap.min.css">
    <style>
    body,html {
        height: 100%;
        margin: 0;
        padding: 0;
        overflow: hidden;
        background: black;
        font-family: "Arial", sans-serif;
    }

    #bg {
        position: fixed;
        top: 0;
        left: 0;
        width: 100%;
        height: 100%;
        object-fit: contain;
    }

    #timer {
        position: absolute;
        top: 10px;
        left: 50%;
        transform: translateX(-50%);
        font-size: 40px;
        color: white;
        text-shadow: 2px 2px 5px black;
        z-index: 10;
    }

    #cat {
        position: absolute;
        bottom: 0;
        width: 120px;
        animation: walk 8s linear infinite alternate;
        z-index: 10;
    }

    @keyframes walk {
        from {left: 0;}
        to {left: calc(100% - 120px);}
    }

    #startBtn {
        position: absolute;
        top: 70px;
        left: 50%;
        transform: translateX(-50%);
        z-index: 10;
    }

    </style>
</head>
<body>
    <img id="bg" src="/static/bg.jpg">
    <div id="timer">25:00</div>
    <button id="startBtn" class="btn btn-success btn-lg">開始番茄鐘🍅</button>
    <img src="/static/cat.gif" id="cat">

<script>
let timerDuration = 25 * 60; // 25 分鐘
let timerInterval;

document.getElementById("startBtn").onclick = function(){
    clearInterval(timerInterval);
    let time = timerDuration;

    timerInterval = setInterval(function(){
        let minutes = Math.floor(time / 60);
        let seconds = time % 60;

        document.getElementById("timer").textContent =
            (minutes < 10 ? '0' : '') + minutes + ":" + (seconds < 10 ? '0' : '') + seconds;

        if(time <= 0){
            clearInterval(timerInterval);
            alert("時間到！休息一下喵～🐱");
        }
        time--;
    }, 1000);
};
</script>
</body>
</html>

```
