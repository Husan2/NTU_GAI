#畫一隻函數蝴蝶
- **使用 Google Colab 的 AI 生成指令，讓 AI 幫助我們畫出蝴蝶的形狀。**
## 過程
    1. `x` 和 `y` 是翅膀的座標
        - 分別使用 `sin(t)` 和 `cos(t)` 來控制形狀的對稱性。

    2. 使用參數 `t` 定義翅膀形狀
        - `t` 是從 `0` 到 `2π` 之間的數據點，控制翅膀的形狀。
    3. 計算蝴蝶翅膀的座標
        - `np.exp(np.cos(t))` 產生一個類似高斯函數的變化，使翅膀具有更自然的形狀。
    4. 調整翅膀的內部細節
        - `2 * np.cos(4 * t)` 這項負的餘弦函數調整翅膀的內部細節，使得它們在特定區域變得較窄或較寬。
    5. 增加曲線的變化
        - `np.sin(t / 12)**5` 是一個微小的修正項，使得蝴蝶翅膀有更多細節。
    6. 繪製對稱的翅膀
        - 透過 `plt.plot(x, y)` 與 `plt.plot(-x, y)` 讓左右翅膀對稱顯示。

```python
butterfly_t = np.linspace(0, 80, 1000)
butterfly_x = np.sin(butterfly_t) * (np.exp(np.cos(butterfly_t)) - 2 * np.cos(4 * butterfly_t))
butterfly_y = np.cos(butterfly_t) * (np.exp(np.cos(butterfly_t)) - 2 * np.cos(4 * butterfly_t))

plt.figure(figsize=(8, 8))
plt.plot(butterfly_x, butterfly_y, color='purple', linewidth=3)
plt.title('Butterfly')
plt.show()
```

# 畫一朵花
- **這張圖是透過 ChatGPT 產生的，因為我想知道 GPT 認為「用函數畫花」會長什麼樣子。**

## 過程
1. **設定花瓣數量與角度**  
   - `angles = np.linspace(0, 2π, petals, endpoint=False)` 設定花瓣角度。

2. **計算每個花瓣的座標**  
   - 使用 `sin` 與 `cos` 設計花瓣形狀，使其環繞中心對稱排列。

3. **繪製多個對稱花瓣**  
   - 利用 `for` 迴圈繪製每個花瓣，使整體看起來像花。

```python
def draw_flower_pattern():
    fig, ax = plt.subplots()

    # 設置花瓣數量與角度
    petals = 12
    angles = np.linspace(0, 2 * np.pi, petals, endpoint=False)

    for angle in angles:
        t = np.linspace(0, np.pi, 100)
        x = 0.3 * np.sin(t) * np.cos(angle) - 0.15 * np.cos(t) * np.sin(angle)
        y = 0.3 * np.sin(t) * np.sin(angle) + 0.15 * np.cos(t) * np.cos(angle)

        ax.plot(x, y, color='brown')
        ax.plot(-x, -y, color='brown')

    ax.set_aspect('equal')
    ax.axis("off")
    plt.show()

draw_flower_pattern()
```

## 目標
透過簡單的幾何形狀 (矩形、圓形、線條) 來描繪一隻黑貓的輪廓。

## 過程
1. **畫貓的身體 (矩形)**  
   - 透過 `plt.fill()` 繪製一個黑色矩形作為身體。

2. **畫貓的耳朵 (三角形)**  
   - 利用 `plt.fill()` 畫兩個三角形，讓貓的輪廓更明顯。

3. **畫貓的眼睛、鼻子與嘴巴**  
   - `plt.scatter()` 畫眼睛，`plt.plot()` 加上嘴巴與細節，使圖案更完整。

4. **畫貓的尾巴 (曲線)**  
   - 使用 `sin` 函數畫出貓的彎曲尾巴。

```python
def draw_cat():
    fig, ax = plt.subplots()

    # 畫貓的身體 (矩形)
    body_x = [-0.5, 0.5, 0.5, -0.5, -0.5]
    body_y = [-0.5, -0.5, 0.5, 0.5, -0.5]
    ax.fill(body_x, body_y, 'k')

    # 畫貓的耳朵 (三角形)
    ear1_x = [-0.5, -0.3, -0.4]
    ear1_y = [0.5, 0.8, 0.5]
    ax.fill(ear1_x, ear1_y, 'k')

    ear2_x = [0.5, 0.3, 0.4]
    ear2_y = [0.5, 0.8, 0.5]
    ax.fill(ear2_x, ear2_y, 'k')

    # 畫眼睛 (兩個圓)
    eye_x = [-0.2, 0.2]
    eye_y = [0.2, 0.2]
    ax.scatter(eye_x, eye_y, s=100, c='w', edgecolors='k')

    # 畫鼻子 (小圓)
    ax.scatter(0, 0, s=40, c='w', edgecolors='k')

    # 畫嘴巴 (小線條)
    mouth_x = [0, 0, 0.05, -0.05]
    mouth_y = [0, -0.1, -0.15, -0.15]
    ax.plot(mouth_x, mouth_y, 'k')

    # 畫尾巴 (曲線)
    tail_t = np.linspace(0, np.pi, 50)
    tail_x = 0.5 + 0.3 * np.cos(tail_t)
    tail_y = -0.3 + 0.4 * np.sin(tail_t)
    ax.plot(tail_x, tail_y, 'k', linewidth=3)

    ax.set_aspect('equal')
    ax.axis("off")
    plt.show()

draw_cat()
```
